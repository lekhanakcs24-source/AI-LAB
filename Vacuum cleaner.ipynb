```python
# Vacuum Cleaner Agent

# Initial state
room_A = input("Enter status of Room A (Clean/Dirty): ")
room_B = input("Enter status of Room B (Clean/Dirty): ")

vacuum = input("Enter vacuum cleaner position (A/B): ")

# Convert input to uppercase
room_A = room_A.upper()
room_B = room_B.upper()
vacuum = vacuum.upper()

print("\nInitial State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position:", vacuum)

# Agent works until both rooms are clean
while room_A == "DIRTY" or room_B == "DIRTY":

    if vacuum == "A":

        if room_A == "DIRTY":
            print("\nVacuum is in Room A")
            print("Action: SUCK")
            room_A = "CLEAN"

        else:
            print("\nRoom A is already clean")
            print("Action: MOVE RIGHT")
            vacuum = "B"

    elif vacuum == "B":

        if room_B == "DIRTY":
            print("\nVacuum is in Room B")
            print("Action: SUCK")
            room_B = "CLEAN"

        else:
            print("\nRoom B is already clean")
            print("Action: MOVE LEFT")
            vacuum = "A"

print("\nFinal State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position:", vacuum)

print("\nGoal achieved!")
print("Both rooms are CLEAN.")
```
