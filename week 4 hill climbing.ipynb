```python
import random

N = 4

def calculate_heuristic(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Check diagonal attacks
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def print_board(state):
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print()


def get_best_neighbor(state):
    best_state = state[:]
    best_h = calculate_heuristic(state)

    for col in range(N):
        for row in range(N):
            if row != state[col]:
                neighbor = state[:]
                neighbor[col] = row

                h = calculate_heuristic(neighbor)

                if h < best_h:
                    best_h = h
                    best_state = neighbor

    return best_state, best_h


def hill_climbing():
    # Randomly place one queen in each column
    state = [random.randint(0, N - 1) for _ in range(N)]

    print("Initial Board:")
    print_board(state)

    while True:
        h = calculate_heuristic(state)

        if h == 0:
            print("Solution Found!")
            print_board(state)
            return state

        neighbor, neighbor_h = get_best_neighbor(state)

        # Restart if stuck in a local minimum
        if neighbor_h >= h:
            print("Local minimum reached. Restarting...\n")
            state = [random.randint(0, N - 1) for _ in range(N)]
        else:
            state = neighbor
            print("Next Board (Heuristic =", neighbor_h, "):")
            print_board(state)


solution = hill_climbing()
print("Queen positions:", solution)
print("Lekhana K 1WN24CS145")
```
