```python

def occurs_check(var, term):
    """Check whether a variable occurs in a term."""
    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, arg) for arg in term)

    return False


def apply_substitution(subst, term):
    """Apply substitutions recursively to a term."""
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    if isinstance(term, tuple):
        return tuple(apply_substitution(subst, arg) for arg in term)

    return term


def unify(t1, t2, subst=None):
    """Unification algorithm."""
    if subst is None:
        subst = {}

    t1 = apply_substitution(subst, t1)
    t2 = apply_substitution(subst, t2)

    # Step 1: If both terms are identical
    if t1 == t2:
        return subst

    # Step 2: If t1 is a variable
    if isinstance(t1, str) and t1.islower():
        if occurs_check(t1, t2):
            return None

        subst[t1] = t2
        return subst

    # Step 3: If t2 is a variable
    if isinstance(t2, str) and t2.islower():
        if occurs_check(t2, t1):
            return None

        subst[t2] = t1
        return subst

    # Step 4: Check predicate/function names
    if isinstance(t1, tuple) and isinstance(t2, tuple):

        if t1[0] != t2[0]:
            return None

        if len(t1) != len(t2):
            return None

        # Step 5: Unify corresponding arguments
        for a, b in zip(t1[1:], t2[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    # Unification failure
    return None


# Example
expr1 = ("P", "x", ("f", "y"))
expr2 = ("P", "a", ("f", "b"))

result = unify(expr1, expr2)

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failure")
```
