```python
import random
import math

N = 8

def heuristic(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def print_board(state):
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print()


def simulated_annealing():
    # Random initial state: one queen per column
    current = [random.randint(0, N - 1) for _ in range(N)]

    T = 1000.0
    cooling_rate = 0.99
    min_temperature = 0.01

    while T > min_temperature:
        h = heuristic(current)

        if h == 0:
            print("Solution Found!")
            print_board(current)
            return current

        # Generate a random neighboring state
        neighbor = current[:]
        col = random.randint(0, N - 1)
        row = random.randint(0, N - 1)

        neighbor[col] = row

        delta = heuristic(neighbor) - h

        # Accept a better state or probabilistically accept a worse one
        if delta < 0:
            current = neighbor
        else:
            probability = math.exp(-delta / T)
            if random.random() < probability:
                current = neighbor

        # Reduce temperature
        T *= cooling_rate

    print("No solution found. Try running again.")
    print("Final Board:")
    print_board(current)
    return current


solution = simulated_annealing()

print("Queen positions:", solution)
print("Lekhana K 1WN24CS145")
```
